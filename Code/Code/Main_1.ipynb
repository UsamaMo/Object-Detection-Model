```python
from pathlib import Path
import sys

# Works from the repository root or Code/Code.
ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "detector.py").is_file()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside the Object-Detection-Model repository.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from detector import catalog, detect, read_image, stitch, encode, evaluate
from IPython.display import Image, display
print(f"Loaded {len(catalog('objects'))} objects and {len(catalog('scenes'))} scenes")
```

## Detect reference objects
Choose any of the supplied scenes, S1 through S20. The displayed image and exported coordinates use a maximum side of 1600 pixels.

```python
scene_path = ROOT / "Scenes" / "S1.png"
scene = read_image(scene_path)
rows, images, keypoint_count = detect(scene, ratio=0.70, min_inliers=7)
print(f"{sum(row['detected'] for row in rows)} objects confirmed; {keypoint_count} scene keypoints")
for row in rows:
    print(row['name'], "detected" if row['detected'] else "not confirmed", "— inliers:", row['inliers'])
display(Image(data=images['detection']))
```

## Inspect features and matches

```python
display(Image(data=images['keypoints']))
display(Image(data=images['O1']))  # Change the reference ID to O1–O15.
```

## Save results

```python
(ROOT / 'Detected_Objects' / f'{scene_path.stem}.jpg').write_bytes(images['detection'])
(ROOT / 'Keypoints' / f'{scene_path.stem}.jpg').write_bytes(images['keypoints'])
for object_path in catalog('objects'):
    (ROOT / 'Matches' / f'{scene_path.stem}_{object_path.stem}.jpg').write_bytes(images[object_path.stem])
```

## Evaluation
Supply manually checked ground truth before computing metrics. The old notebooks included inconsistent labels and objects missing from the library; they are not used as verified ground truth.

```python
expected = None  # Replace with a list such as ["O1", "O3"] after checking the scene.
if expected is None:
    print("Set expected to the reference IDs visible in this scene to calculate metrics.")
else:
    print(evaluate([row['id'] for row in rows if row['detected']], expected,
                   [p.stem for p in catalog('objects')]))
```

## Panorama
Stitching requires overlapping views from similar camera positions. Failure is reported without crashing the notebook.

```python
try:
    panorama = stitch([read_image(ROOT / 'Scenes' / name, 1200) for name in ['S1.png', 'S2.png']])
    output = encode(panorama)
    (ROOT / 'Detected_Objects' / 'panorama.jpg').write_bytes(output)
    display(Image(data=output))
except ValueError as exc:
    print(exc)
```
